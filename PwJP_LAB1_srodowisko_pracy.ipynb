## 3. Weryfikacja katalogu roboczego i plików

W kolejnych zadaniach program może korzystać z plików danych. Zanim zmienisz kod, warto sprawdzić, z jakiego katalogu jest uruchamiany i jakie pliki są dostępne.


```python
from pathlib import Path

katalog = Path.cwd()
print("Katalog roboczy:", katalog)
print("Pierwsze dostępne elementy:")
for element in list(katalog.iterdir())[:10]:
    print("-", element.name)
```

### Interpretacja
