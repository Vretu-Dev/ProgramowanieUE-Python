## Zadanie 1: Klasa Wektor (płaszczyzna kartezjańska)

**Polecenie:**
Napisz klasę `Wektor` reprezentującą wektor na płaszczyźnie kartezjańskiej. Klasa ta powinna posiadać dwie składowe „prywatne” – współrzędne `(x, y)` oraz następujące metody publiczne:
* Konstruktor z parametrami ustalającymi współrzędne `x` i `y` oraz wartościami domyślnymi pozwalającymi na stworzenie wektora zerowego,
* Metodę `dlugosc(self)` zwracającą długość wektora,
* Metodę `normalizuj(self)` zwracającą nowy wektor równy znormalizowanemu bieżącemu wektorowi,
* Metodę `wyswietl(self)` wypisującą łańcuch opisujący wektor oraz jego długość (np. `Wektor [1.5, 0] o długości 1.5`).

**Przeciążenie operatorów:**
* Dodawanie `__add__(self, w)` i odejmowanie `__sub__(self, w)`,
* Dodawanie i odejmowanie w miejscu modyfikujące `self`: `__iadd__(self, w)` oraz `__isub__(self, w)`,
* Konwersja na napis `__str__(self)` (format np. `[1.20, -3.00]`),
* Mnożenie przez skalar `__mul__(self, a)` oraz mnożenie prawostronne `__rmul__(self, a)` dla zapisu `a * wektor`.


```python
import math

class Wektor:
    def __init__(self, x=0.0, y=0.0):
        # Pola prywatne
        self.__x = float(x)
        self.__y = float(y)

    def dlugosc(self):
        # Długość wektora wyznaczana ze wzoru sqrt(x^2 + y^2)
        return math.sqrt(self.__x**2 + self.__y**2)

    def normalizuj(self):
        # Zwraca nowy wektor o długości 1 (znormalizowany)
        d = self.dlugosc()
        if d == 0:
            return Wektor(0.0, 0.0)
        return Wektor(self.__x / d, self.__y / d)

    def wyswietl(self):
        # Wypisuje i zwraca łańcuch opisujący wektor oraz jego długość
        tekst = f"Wektor {self} o długości {self.dlugosc()}"
        print(tekst)
        return tekst

    def __add__(self, w):
        # Operator dodawania dwóch wektorów: self + w
        return Wektor(self.__x + w.__x, self.__y + w.__y)

    def __sub__(self, w):
        # Operator odejmowania dwóch wektorów: self - w
        return Wektor(self.__x - w.__x, self.__y - w.__y)

    def __iadd__(self, w):
        # Operator dodawania w miejscu: self += w
        self.__x += w.__x
        self.__y += w.__y
        return self

    def __isub__(self, w):
        # Operator odejmowania w miejscu: self -= w
        self.__x -= w.__x
        self.__y -= w.__y
        return self

    def __mul__(self, a):
        # Mnożenie wektora przez skalar: wektor * a
        return Wektor(self.__x * a, self.__y * a)

    def __rmul__(self, a):
        # Mnożenie skalara przez wektor: a * wektor
        return self.__mul__(a)

    def __str__(self):
        # Zwraca reprezentację tekstową wektora w formacie [x.xx, y.xx]
        # Zabezpieczenie przed wyświetlaniem -0.00
        x = 0.0 if abs(self.__x) < 1e-9 else self.__x
        y = 0.0 if abs(self.__y) < 1e-9 else self.__y
        return f"[{x:.2f}, {y:.2f}]"


# Przykład zastosowania klasy Wektor:
w1 = Wektor(2, 4)
w2 = Wektor(1.5)
print("Wektor w1:", w1, "wektor w2:", w2)
print("Dł. w1 =", w1.dlugosc(), "dł. w2 =", w2.dlugosc())
print("w1+w2 =", w1 + w2)
print("w1-w2 =", w1 - w2)
print("w1*2 =", w1 * 2)
print("-3*w2 =", -3 * w2)
print("w1*2-w2 =", w1 * 2 - w2)
print("w1 po normalizacji =", w1.normalizuj())
print("w2 po normalizacji =", w2.normalizuj())
w1.wyswietl()
w2.wyswietl()
```

## Zadanie 2: Obsługa zamówień (ElementZamowienia oraz Zamowienie)

**Polecenie:**
Napisz program do obsługi zamówień składający się z klas `ElementZamowienia` oraz `Zamowienie`:

1. **`ElementZamowienia`** – reprezentuje pojedynczą pozycję zamówienia:
   * Pola prywatne: `nazwa`, `cena`, `liczbaSztuk`,
   * Konstruktor z parametrami przypisującymi wartości początkowe,
   * Przeciążona metoda `__str__()` zwracająca opis elementu (np. `Chleb 4.00 zł, 2 szt., łącznie 8.00 zł`),
   * Metoda `obliczRabat()` wyznaczająca rabat: jeśli zamówiono co najmniej 5 sztuk danego towaru, rabat wynosi 10%, w przeciwnym razie 0%,
   * Metoda `obliczKoszt()` zwracająca łączny koszt pozycji po uwzględnieniu rabatu.

2. **`Zamowienie`** – zarządza całością zamówienia:
   * Pola prywatne: `elementy` (lista obiektów `ElementZamowienia`), `rozmiar` (aktualna liczba pozycji), `maksRozmiar` (maks. liczba pozycji),
   * Konstruktor z parametrem określającym maksymalną liczbę elementów,
   * Metoda `dodaj(self, elZam)` dodająca pozycję do zamówienia (zwraca `True` w przypadku sukcesu lub `False`, gdy przekroczono maks. rozmiar),
   * Metoda `obliczKoszt(self)` zwracająca całkowity koszt zamówienia,
   * Metoda `pisz(self)` wypisująca na ekranie listę pozycji, koszt całkowity oraz naliczony rabat.


```python
class ElementZamowienia:
    def __init__(self, nazwa, cena, liczbaSztuk):
        # Pola prywatne
        self.__nazwa = str(nazwa)
        self.__cena = float(cena)
        self.__liczbaSztuk = int(liczbaSztuk)

    def obliczRabat(self):
        # Rabat wynosi 10%, jeśli klient zamówił co najmniej 5 sztuk towaru
        if self.__liczbaSztuk >= 5:
            return self.__cena * self.__liczbaSztuk * 0.10
        return 0.0

    def obliczKoszt(self):
        # Łączny koszt pozycji z uwzględnieniem rabatu
        return (self.__cena * self.__liczbaSztuk) - self.obliczRabat()

    def __str__(self):
        # Reprezentacja pozycji, np. "Chleb 4.00 zł, 2 szt., łącznie 8.00 zł"
        return f"{self.__nazwa} {self.__cena:.2f} zł, {self.__liczbaSztuk} szt., łącznie {self.obliczKoszt():.2f} zł"


class Zamowienie:
    def __init__(self, maksRozmiar=10):
        # Pola prywatne
        self.__maksRozmiar = int(maksRozmiar)
        self.__elementy = []
        self.__rozmiar = 0

    def dodaj(self, elZam):
        # Dodaje element do zamówienia, jeśli nie przekroczono maksRozmiar
        if self.__rozmiar < self.__maksRozmiar:
            self.__elementy.append(elZam)
            self.__rozmiar += 1
            return True
        return False

    def obliczKoszt(self):
        # Całkowity koszt wszystkich pozycji zamówienia
        return sum(el.obliczKoszt() for el in self.__elementy)

    def pisz(self):
        # Wypisuje podsumowanie zamówienia: listę pozycji, łączny koszt i rabat
        print("Zamowienie:")
        for i, el in enumerate(self.__elementy, 1):
            print(f"{i}. {el}")

        koszt = self.obliczKoszt()
        rabat = sum(el.obliczRabat() for el in self.__elementy)

        # Formatowanie liczb (np. 37.5 zamiast 37.50, 2 zamiast 2.0)
        koszt_str = f"{int(koszt)}" if isinstance(koszt, float) and koszt.is_integer() else f"{koszt:g}"
        rabat_str = f"{int(rabat)}" if isinstance(rabat, float) and rabat.is_integer() else f"{rabat:g}"

        print(f"Koszt całkowity: {koszt_str} zł")
        print(f"Naliczony rabat: {rabat_str} zł")


# Przykład zastosowania klas:
z = Zamowienie(10)
z.dodaj(ElementZamowienia("Chleb", 4.0, 2))
z.dodaj(ElementZamowienia("Mleko", 2.5, 1))
z.dodaj(ElementZamowienia("Cukier", 4.0, 5))
z.dodaj(ElementZamowienia("Puszka", 9.0, 1))
z.pisz()
```
