## Zadanie 1: Kalkulator BMI

**Polecenie:**
Napisz program obliczający wskaźnik masy ciała **BMI** ($BMI = \frac{\text{waga}}{\text{wzrost}^2}$). Program pobiera wagę (w kg) oraz wzrost (w cm), a następnie wyświetla wartość BMI oraz diagnozę (niedowaga, waga prawidłowa, nadwaga).


```python
waga = float(input("Podaj waga: "))
wzrost = float(input("Podaj wzrost: "))

wzrost = wzrost / 100

bmi = waga / (wzrost * wzrost)

if bmi >= 18.5 and bmi <= 24.9:
    print(f"Waga prawidłowa | BMI = {bmi}")
elif bmi > 24.9:
    print(f"Nadwaga | BMI = {bmi}")
else:
    print(f"Niedowaga | BMI = {bmi}")
```

## Zadanie 7: Gra 'Za dużo, za mało'

**Polecenie:**
Napisz grę *'Za dużo, za mało'*. Użytkownik wybiera poziom trudności:
* **Easy (1):** Przedział 1–50, 10 szans
* **Medium (2):** Przedział 1–100, 15 szans
* **Hard (3):** Przedział 1–200, 20 szans

Program losuje liczbę, a użytkownik próbuje ją odgadnąć. Po każdej próbie program podpowiada, czy wylosowana liczba jest większa, czy mniejsza od podanej.


```python
import random

print(f"Gra 'Za dużo, za mało'\n"
      f"Podaj poziom trudności:\n"
      f"Easy (1) - Przedział 1-50 | 10 Szans\n"
      f"Medium (2) - Przedział 1-100 | 15 Szans\n"
      f"Hard (3) - Przedział 1-200 | 20 Szans")

poziom = 0
szansa = 0

while poziom not in (1, 2 ,3):
    poziom = int(input("Podaj poziom: "))

    if poziom == 1:
        losowa = random.randint(1, 50)
        szansa = 10
    elif poziom == 2:
        losowa = random.randint(1, 100)
        szansa = 15
    elif poziom == 3:
        losowa = random.randint(1, 200)
        szansa = 20

while szansa > 0:
    liczba = int(input(f"(Pozostałe szanse) {szansa}\nPodaj liczbę: "))
    szansa -= 1

    if losowa == liczba:
        print(f"Wygrałeś")
        break
    if losowa > liczba:
        print(f"Wylosowana liczba jest większa niż podana")
    else:
        print(f"Wylosowana liczba jest mniejsza niż podana")

else:
    print(f"Przegrałeś")
```

## Zadanie 8: Analiza cyfr liczby

**Polecenie:**
Napisz program, który pobiera liczbę całkowitą i oblicza sumę jej cyfr oraz średnią arytmetyczną cyfr parzystych i nieparzystych, odpowiednio zabezpieczając program przed dzieleniem przez zero.


```python
liczba = input("Podaj liczbę: ")

suma = 0
parzyste = []
nieparzyste = []

for znak in liczba:
    if not znak.isdigit():
        continue
    cyfra = int(znak)
    suma += cyfra
    if cyfra % 2 == 0:
        parzyste.append(cyfra)
    else:
        nieparzyste.append(cyfra)

sredniaParzyste = sum(parzyste) / len(parzyste) if parzyste else "brak" # if warunek ? true : false
sredniaNieparzyste = sum(nieparzyste) / len(nieparzyste) if nieparzyste else "brak"

print("=" * 50)
print(f"Suma cyfr liczby {suma}")
print(f"Średnia arytmetyczna liczb parzystych {sredniaParzyste}")
print(f"Średnia arytmetyczna liczb nieparzystych {sredniaNieparzyste}")
print("=" * 50)
```

## Zadanie 9: Dzielniki, liczba pierwsza i doskonała

```python
from copy import deepcopy

liczba = int(input("Podaj liczba: "))

isDoskonala = True
isPierwsza = True
dzielniki = []

for i in range(1, liczba + 1):
    if liczba % i == 0:
        dzielniki.append(i)

if len(dzielniki) != 2:
    isPierwsza = False

doskonale = deepcopy(dzielniki)
doskonale.pop()

sumaDoskonale = sum(doskonale)

if sumaDoskonale != liczba:
    isDoskonala = False

print(f"Dzielniki liczby {liczba} to {dzielniki}\n"
      f"Czy pierwsza: {isPierwsza}\n"
      f"Czy doskonała: {isDoskonala}")
```

## Zadanie 10: Liczba pierwsza i liczby pierwsze w zakresie [1, n]

**Polecenie:**
Napisz program, który sprawdza, czy podana liczba całkowita $n$ ($n > 1$) jest liczbą pierwszą. Dodaj funkcjonalność wyświetlającą wszystkie liczby pierwsze w zakresie od 1 do $n$.


```python

def czy_pierwsza(liczba):
    if liczba < 2:
        return False

    for d in range(2, int(liczba ** 0.5) + 1):
        if liczba % d == 0:
            return False
    return True

while True:
    try:
        n = int(input("Podaj liczbę całkowitą n (n > 1): "))
        if n > 1:
            break
        print("Liczba musi być większa niż 1! Spróbuj ponownie.")
    except ValueError:
        print("Nieprawidłowa wartość! Wprowadź liczbę całkowitą.")

if czy_pierwsza(n):
    print(f"\nLiczba {n} JEST liczbą pierwszą.")
else:
    print(f"\nLiczba {n} NIE JEST liczbą pierwszą.")

pierwsze_w_zakresie = [x for x in range(2, n + 1) if czy_pierwsza(x)]

print(f"\nLiczby pierwsze w zakresie od 1 do {n} (łącznie {len(pierwsze_w_zakresie)}):")
print(pierwsze_w_zakresie)
```

## Zadanie 11: Rozwiązywanie układu równań liniowych metodą eliminacji Gaussa

**Polecenie:**
Napisz program, który rozwiązuje układ równań liniowych dwóch zmiennych metodą eliminacji Gaussa. Program powinien pobierać współczynniki równań od użytkownika, sprawdzać poprawność układu oraz wyświetlać rozwiązanie (lub informować o jego braku / nieskończenie wielu rozwiązaniach).


```python
# Układ
# a1 * x + b1 * y = c1
# a2 * x + b2 * y = c2

def pobierz_wspolczynnik(nazwa):
    while True:
        try:
            return float(input(f"Podaj współczynnik {nazwa}: "))
        except ValueError:
            print("Błąd: wprowadź poprawną liczbę!")

print("=" * 55)
print("  ROZWIĄZYWANIE UKŁADU RÓWNAŃ METODĄ ELIMINACJI GAUSSA")
print("          a1*x + b1*y = c1")
print("          a2*x + b2*y = c2")
print("=" * 55)

print("\n--- Równanie 1 ---")
a1 = pobierz_wspolczynnik("a1")
b1 = pobierz_wspolczynnik("b1")
c1 = pobierz_wspolczynnik("c1")

print("\n--- Równanie 2 ---")
a2 = pobierz_wspolczynnik("a2")
b2 = pobierz_wspolczynnik("b2")
c2 = pobierz_wspolczynnik("c2")

# Macierz rozszerzoną układu [A|B]
# Wiersz 1: [a1, b1, c1]
# Wiersz 2: [a2, b2, c2]
R1 = [a1, b1, c1]
R2 = [a2, b2, c2]

print("\nPostać początkowa macierzy rozszerzonej [A | B]:")
print(f"[{R1[0]:8.2f} {R1[1]:8.2f} | {R1[2]:8.2f}]")
print(f"[{R2[0]:8.2f} {R2[1]:8.2f} | {R2[2]:8.2f}]")

EPSILON = 1e-9

# 1. Wybór elementu podstawowego (pivoting) - jeśli w pierwszym wierszu a1 jest zerem, zamieniamy wiersze
if abs(R1[0]) < EPSILON and abs(R2[0]) >= EPSILON:
    print("\n[Operacja] Zamiana wiersza 1 z wierszem 2 (aby uniknąć zera na przekątnej)")
    R1, R2 = R2, R1

# 2. Eliminacja Gaussa
if abs(R1[0]) < EPSILON and abs(R2[0]) < EPSILON:
    # Współczynniki przy x w obu równaniach to 0
    if abs(R1[1]) < EPSILON and abs(R2[1]) < EPSILON:
        if abs(R1[2]) < EPSILON and abs(R2[2]) < EPSILON:
            print("\nWynik: Układ tożsamościowy (nieoznaczony) - nieskończenie wiele rozwiązań.")
        else:
            print("\nWynik: Układ sprzeczny - brak rozwiązań.")
    else:
        print("\nZmienna x nie występuje w układzie. Równania zawierają tylko y.")
        # Sprawdzamy wyznaczenie y
        y1 = R1[2] / R1[1] if abs(R1[1]) >= EPSILON else None
        y2 = R2[2] / R2[1] if abs(R2[1]) >= EPSILON else None
        if y1 is not None and y2 is not None:
            if abs(y1 - y2) < EPSILON:
                print(f"Rozwiązanie: y = {y1:.2f}, x należy do zbioru liczb rzeczywistych.")
            else:
                print("Wynik: Układ sprzeczny - brak rozwiązań.")
        elif y1 is not None:
            print(f"Rozwiązanie: y = {y1:.2f}, x należy do zbioru liczb rzeczywistych.")
        else:
            print(f"Rozwiązanie: y = {y2:.2f}, x należy do zbioru liczb rzeczywistych.")
else:
    # Wyzerowanie współczynnika przy x w drugim wierszu
    wspolczynnik_m = R2[0] / R1[0]
    print(f"\n[Operacja] Wiersz 2 = Wiersz 2 - ({wspolczynnik_m:.2f}) * Wiersz 1")

    R2[0] = 0.0
    R2[1] = R2[1] - wspolczynnik_m * R1[1]
    R2[2] = R2[2] - wspolczynnik_m * R1[2]

    print("\nMacierz po eliminacji (postać schodkowa):")
    print(f"[{R1[0]:8.2f} {R1[1]:8.2f} | {R1[2]:8.2f}]")
    print(f"[{R2[0]:8.2f} {R2[1]:8.2f} | {R2[2]:8.2f}]")

    # 3. Analiza rozwiązań i postępowanie wsteczne (back-substitution)
    if abs(R2[1]) < EPSILON:
        if abs(R2[2]) < EPSILON:
            print("\nWynik: Układ nieoznaczony - nieskończenie wiele rozwiązań (zależnych od jednego parametru).")
        else:
            print("\nWynik: Układ sprzeczny (0 = const != 0) - brak rozwiązań.")
    else:
        # Dokładnie jedno rozwiązanie
        y = R2[2] / R2[1]
        x = (R1[2] - R1[1] * y) / R1[0]
        print("\n" + "=" * 35)
        print("  UKŁAD OZNACZONY (JEDNO ROZWIĄZANIE)")
        print(f"  x = {x:.4f}")
        print(f"  y = {y:.4f}")
        print("=" * 35)
```

## Zadanie 12: Algorytm Euklidesa i tożsamość Bézouta

**Polecenie:**
Napisać program, który implementuje algorytm Euklidesa do znalezienia największego wspólnego dzielnika (NWD) dwóch liczb oraz rozszerzony algorytm Euklidesa do znalezienia współczynników wyrażenia Bézouta ($a \cdot x + b \cdot y = \text{NWD}(a, b)$). Program powinien wyświetlać pełny przebieg obliczeń krok po kroku.


```python

# NWD(a, b) oraz współczynniki całkowite x, y:
# a * x + b * y = NWD(a, b)

def pobierz_liczbe(etykieta):
    while True:
        try:
            wartosc = int(input(f"Podaj {etykieta} (liczbę całkowitą dodatnią): "))
            if wartosc > 0:
                return wartosc
            print("Liczba musi być większa od zera!")
        except ValueError:
            print("Wprowadź poprawną liczbę całkowitą!")

print("=" * 65)
print("      ALGORYTM EUKLIDESA I ROZSZERZONY ALGORYTM EUKLIDESA")
print("          Wyznaczanie NWD(a, b) oraz tożsamości Bézouta:")
print("                     a * x + b * y = NWD(a, b)")
print("=" * 65)

a = pobierz_liczbe("liczbę a")
b = pobierz_liczbe("liczbę b")

# 1. Klasyczny algorytm Euklidesa z pełnym przebiegiem
print("\n" + "-" * 65)
print(" CZĘŚĆ 1: KLASYCZNY ALGORYTM EUKLIDESA (Kolejne dzielenia z resztą)")
print("-" * 65)

temp_a, temp_b = a, b
krok = 1

while temp_b != 0:
    q = temp_a // temp_b
    r = temp_a % temp_b
    print(f"Krok {krok:2d}:  {temp_a:6d} = {q:4d} * {temp_b:6d} + {r:6d}   (reszta = {r})")
    temp_a, temp_b = temp_b, r
    krok += 1

nwd = temp_a
print(f"\n=> Największy Wspólny Dzielnik: NWD({a}, {b}) = {nwd}")

# 2. Rozszerzony algorytm Euklidesa (tabelaryczny)
print("\n" + "-" * 65)
print(" CZĘŚĆ 2: ROZSZERZONY ALGORYTM EUKLIDESA (Współczynniki Bézouta)")
print(" Wyznaczamy ciągi r, x, y takie, że w każdym kroku: r_i = x_i * a + y_i * b")
print("-" * 65)

# Stan początkowy:
# r0 = a, x0 = 1, y0 = 0  (bo a = 1*a + 0*b)
# r1 = b, x1 = 0, y1 = 1  (bo b = 0*a + 1*b)
r_prev, r_curr = a, b
x_prev, x_curr = 1, 0
y_prev, y_curr = 0, 1

print(f"{'Krok':^6} | {'Iloraz q':^10} | {'Reszta r':^10} | {'x':^10} | {'y':^10} | {'Równanie: r = x*a + y*b'}")
print("-" * 75)
print(f"{'Start':^6} | {'-':^10} | {r_prev:^10} | {x_prev:^10} | {y_prev:^10} | {r_prev} = ({x_prev})*{a} + ({y_prev})*{b}")
print(f"{'Start':^6} | {'-':^10} | {r_curr:^10} | {x_curr:^10} | {y_curr:^10} | {r_curr} = ({x_curr})*{a} + ({y_curr})*{b}")

krok = 1
while r_curr != 0:
    q = r_prev // r_curr
    r_next = r_prev - q * r_curr
    x_next = x_prev - q * x_curr
    y_next = y_prev - q * y_curr

    print(f"{krok:^6} | {q:^10} | {r_next:^10} | {x_next:^10} | {y_next:^10} | {r_next} = ({x_next})*{a} + ({y_next})*{b}")

    r_prev, r_curr = r_curr, r_next
    x_prev, x_curr = x_curr, x_next
    y_prev, y_curr = y_curr, y_next
    krok += 1

x, y = x_prev, y_prev

print("\n" + "=" * 65)
print(" PODSUMOWANIE I WERYFIKACJA WYNIKÓW:")
print("=" * 65)
print(f" NWD({a}, {b}) = {nwd}")
print(f" Współczynniki Bézouta: x = {x}, y = {y}")
print(f" Równanie Bézouta:      {a} * ({x}) + {b} * ({y}) = {a * x + b * y}")
print(f" Zgodność z NWD:        {a * x + b * y} == {nwd} -> {a * x + b * y == nwd}")
print("=" * 65)
```
