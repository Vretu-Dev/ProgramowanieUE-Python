```python
tekst = input("Podaj ciąg znaków: ")

ostatni_znak = tekst[-1]
licznik = 0

for znak in tekst:
    if znak == ostatni_znak:
      licznik += 1

print(f"Ostatni znak '{ostatni_znak}' występuje w ciągu {licznik} razy.")

liczba_wystapien = tekst.count(ostatni_znak)
print(f"Ostatni znak '{ostatni_znak}' występuje w ciągu {liczba_wystapien} razy.")
```

```python
tekst = input("Podaj ciąg znaków: ")

isPalindrom = True

for i in range(len(tekst) // 2):
  if tekst[i] != tekst[-i - 1]:
    isPalindrom = False
    break

print(f"Słowo: {tekst} | Palindrom: {isPalindrom}")
```

```python
tekst = input("Podaj ciąg znaków: ")

suma = 0

for znak in tekst:
    if znak.isdigit():
        suma += int(znak)

print(f"Suma: {suma}")
```

```python
alfabet = "abcdefghijklmnopqrstuvwxyz"

tekst = input("Podaj łańcuch znaków do zaszyfrowania: ")
n = int(input("Podaj przesunięcie: "))

wynik = ""

for znak in tekst:
  if znak == " ":
    wynik += " "
  else:
    indeks = alfabet.index(znak)
    nowy_indeks = (indeks + n) % len(alfabet)
    wynik += alfabet[nowy_indeks]

print(f"Zaszyfrowany tekst: {wynik}")
```

```python
def strToInt(tekst: str) -> int:
  if not tekst:
    return 0

  indeks = 0
  znak = 1

  # Znak na poczatku +/-
  if tekst[0] == "-":
    znak = -1
    indeks = 1
  elif tekst[0] == "+":
    indeks = 1

  # Liczba
  liczba = 0
  ma_cyfry = False

  while indeks < len(tekst) and "0" <= tekst[indeks] <= "9":
    liczba = liczba * 10 + (ord(tekst[indeks]) - ord("0"))
    ma_cyfry = True
    indeks += 1

  # nie było cyfry
  if not ma_cyfry:
    return 0

  # ('e' lub 'E') tylko dla dodatnich wykładników
  if indeks < len(tekst) and tekst[indeks] in ("e", "E"):
    j = indeks + 1
    if j < len(tekst) and tekst[j] == "+":
      j += 1

    wykladnik = 0
    ma_wykladnik = False

    while j < len(tekst) and "0" <= tekst[j] <= "9":
      wykladnik = wykladnik * 10 + (ord(tekst[j]) - ord("0"))
      ma_wykladnik = True
      j += 1

    if ma_wykladnik:
      liczba *= 10**wykladnik

  return znak * liczba

przyklady = [
    "-13krowa",
    "+12",
    "0001",
    "991-234-23",
    "+zonk",
    "",
    "-12e5",
    "-12e-5",
]
for p in przyklady:
  print(f'strToInt("{p}") -> {strToInt(p)}')
```

```python
def czyAnagram(t1: str, t2: str) -> bool:
  licznik = {}

  for znak in t1.lower():
    if "a" <= znak <= "z":
      licznik[znak] = licznik.get(znak, 0) + 1

  for znak in t2.lower():
    if "a" <= znak <= "z":
      if znak not in licznik:
        return False
      licznik[znak] -= 1
      if licznik[znak] == 0:
        del licznik[znak]

  return len(licznik) == 0

print(czyAnagram("kolej", "olejk"))  # True
print(czyAnagram("kolej", "kole"))  # False
print(czyAnagram("kolej", "K O L E J"))  # True
print(czyAnagram("Gregory House", "Huge ego, sorry"))  # True
```

```python
def HTMLColor2RGB(color: str) -> list[int]:
  r = int(color[1:3], 16)
  g = int(color[3:5], 16)
  b = int(color[5:7], 16)
  return [r, g, b]


print(HTMLColor2RGB("#FF0050"))  # [255, 0, 80]
print(HTMLColor2RGB("#001020"))  # [0, 16, 32]
```
