1. Napisz program, w którym przygotowane są dwie listy, pierwsz o wielkości n
(podanej przez użytkownka), druga o wielkości m (podanej przez użytkownika).
Następnie listy powinny być wypełnione wartościami losowymi z zakresu [1,10].
Wyświetl listę elementów, które występują w obu listach (czyli ich przecięcie).
Użyj zbiorów do rozwiązania tego problemu. (zbiory)

```python
import random

n = int(input("Podaj rozmiar listy: "))
m = int(input("Podaj liczba: "))

tabN = [random.randint(1, 10) for _ in range(n)]
tabM = [random.randint(1, 10) for _ in range(n)]

setN = set(tabN)
setM = set(tabM)

print(list(setN & setM))
```

2. Napisz program, w którym wczytywany jest ciąg słów (każde słowo, jako kolejny element listy). Następnie, używając zbiorów i słowników,
oblicz częstość występowania każdego słowa w tym ciągu. (zbiory)

```python
tekst = input("Wprowadź tekst: ")

listaSlow = tekst.lower().split()

licznik = {}

for i in range(len(listaSlow)):
    if not licznik.__contains__(listaSlow[i]):
        licznik[listaSlow[i]] = 1
    else:
        licznik[listaSlow[i]] += 1

for k,v in licznik.items():
    print(f"{k}: {v}")
```

3. Stwórz listę 20 losowych wartości z zakresu [0,10]. Następnie (przy zastosowaniu
zbiorów) wyświetl listę wszystkich podzbiorów wartości tych liczb. (zbiory)

```python
from itertools import combinations
import random

tabN = [random.randint(1, 10) for _ in range(20)]

setN = set(tabN)

podzbiory = []

for r in range(len(setN) + 1):
    for kombinacja in combinations(setN, r):
        podzbiory.append(set(kombinacja))

for p in podzbiory:
    print(p)
```

4. Z zastosowaniem słowników napisać funkcję, której parametrem będzie zmienna
zawierająca łańcuch znaków. W funkcji należy policzyć częstość występowania
każdego znaku. Funkcja powinna zwrócić słownik zawierających wszystkie znaki
(jako klucze) i ich wystąpienia (wartość).

```python
tekst = input("Wprowadź tekst: ")

listaZnakow = list(tekst)

def LiczenieZnakow(lista: list) -> dict[str, int]:
    licznik = {}
    for i in range(len(lista)):
        if not licznik.__contains__(lista[i]):
            licznik[lista[i]] = 1
        else:
            licznik[lista[i]] += 1

    return licznik

for k,v in LiczenieZnakow(listaZnakow).items():
    print(f"{k}: {v}")
```

5. Z zastosowaniem słowników należy napisać funkcję, dzięki które można dokonać
zmiany liczby podanej przez parametr (w postaci int) na zapis słowny tej liczby.
Można założyć, że liczba zawsze będzie mniejsza niż milion.

```python
def liczba_na_slowa(liczba: int) -> str:
    if not (0 <= liczba < 1_000_000):
        raise ValueError("Liczba musi mieścić się w przedziale od 0 do 999 999.")

    if liczba == 0:
        return "zero"

    jednosci = {
        1: "jeden", 2: "dwa", 3: "trzy", 4: "cztery", 5: "pięć",
        6: "sześć", 7: "siedem", 8: "osiem", 9: "dziewięć"
    }
    nastki = {
        10: "dziesięć", 11: "jedenaście", 12: "dwanaście", 13: "trzynaście",
        14: "czternaście", 15: "piętnaście", 16: "szesnaście",
        17: "siedemnaście", 18: "osiemnaście", 19: "dziewiętnaście"
    }
    dziesiatki = {
        20: "dwadzieścia", 30: "trzydzieści", 40: "czterdzieści",
        50: "pięćdziesiąt", 60: "sześćdziesiąt", 70: "siedemdziesiąt",
        80: "osiemdziesiąt", 90: "dziewięćdziesiąt"
    }
    setki = {
        100: "sto", 200: "dwieście", 300: "trzysta", 400: "czterysta",
        500: "pięćset", 600: "sześćset", 700: "siedemset",
        800: "osiemset", 900: "dziewięćset"
    }

    def konwertuj_trojke(n: int) -> list[str]:
        slowa = []
        # Setki
        s = (n // 100) * 100
        if s in setki:
            slowa.append(setki[s])

        # Dziesiątki i jedności
        reszta = n % 100
        if reszta in nastki:
            slowa.append(nastki[reszta])
        else:
            d = (reszta // 10) * 10
            if d in dziesiatki:
                slowa.append(dziesiatki[d])
            j = reszta % 10
            if j in jednosci:
                slowa.append(jednosci[j])
        return slowa

    czesci = []
    tysiace = liczba // 1000
    jednostki = liczba % 1000

    if tysiace > 0:
        if tysiace == 1:
            czesci.append("tysiąc")
        else:
            czesci.extend(konwertuj_trojke(tysiace))
            ostatnia_cyfra = tysiace % 10
            ostatnie_dwie = tysiace % 100

            # Odmiana słowa "tysiąc"
            if ostatnia_cyfra in (2, 3, 4) and ostatnie_dwie not in (12, 13, 14):
                czesci.append("tysiące")
            else:
                czesci.append("tysięcy")

    if jednostki > 0:
        czesci.extend(konwertuj_trojke(jednostki))

    return " ".join(czesci)


print(liczba_na_slowa(0))
print(liczba_na_slowa(14))
print(liczba_na_slowa(125))
print(liczba_na_slowa(1001))
print(liczba_na_slowa(24312))
print(liczba_na_slowa(999999))
```

6. Z zastosowaniem słowników należy napisać program „fiszki”. W wyniku działania
programu powinno być możliwe dodawanie słowa i jego znaczenia, pamiętając że
tłumaczone słowo może wystąpić tylko raz. Następnie program losowo wyświetla
słowa pytając o ich znaczenie i zlicza liczbę poprawnych oraz błędnych odpowiedzi.

```python
import random

def program_fiszki():
    fiszki = {}

    print("=== ETAP 1: DODAWANIE SŁÓWEK ===")
    print("Wpisz 'stop' jako słowo, aby zakończyć wprowadzanie.\n")

    while True:
        slowo = input("Podaj słowo (język obcy): ").strip()
        if slowo.lower() == "stop":
            break

        if not slowo:
            print("Słowo nie może być puste.")
            continue

        if slowo in fiszki:
            print(f"Błąd: Słowo '{slowo}' już istnieje w bazie fiszek!")
            continue

        znaczenie = input(f"Podaj znaczenie dla '{slowo}': ").strip()
        fiszki[slowo] = znaczenie
        print(f"Dodano: {slowo} -> {znaczenie}\n")

    if not fiszki:
        print("Nie dodano żadnych słówek. Koniec programu.")
        return

    print("\n=== ETAP 2: QUIZ ===")
    print("Wpisz 'koniec', aby zakończyć test.\n")

    poprawne = 0
    bledne = 0
    lista_slow = list(fiszki.keys())

    while True:
        wylosowane_slowo = random.choice(lista_slow)
        poprawne_znaczenie = fiszki[wylosowane_slowo]

        odpowiedz = input(f"Jakie jest znaczenie słowa '{wylosowane_slowo}'? ").strip()

        if odpowiedz.lower() == "koniec":
            break

        if odpowiedz.lower() == poprawne_znaczenie.lower():
            print("Dobrze!\n")
            poprawne += 1
        else:
            print(f"Błąd! Prawidłowa odpowiedź to: '{poprawne_znaczenie}'\n")
            bledne += 1

    suma = poprawne + bledne
    print("\n=== PODSUMOWANIE WYNIKÓW ===")
    print(f"Liczba pytań: {suma}")
    print(f"Poprawne odpowiedzi: {poprawne}")
    print(f"Błędne odpowiedzi:   {bledne}")
    if suma > 0:
        skutecznosc = (poprawne / suma) * 100
        print(f"Skuteczność: {skutecznosc:.1f}%")

if __name__ == "__main__":
    program_fiszki()
```
